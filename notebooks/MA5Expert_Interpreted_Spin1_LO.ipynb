```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from ma5_expert.cutflow import Collection
from ma5_expert.histogram.reader import Collection

def df(fileLoc):
    # Instantiate a Collection object with the provided file location
    data = Collection(original_file=fileLoc)
    
    # Assuming you want to extract specific information from the histograms
    histograms = []
    for histo_name in data.histo_names:
        xbins, bins, weights = data.get_histogram(histo_name)
        histograms.append({
            "binMin": xbins[:-1],
            "binMax": xbins[1:],
            "value": weights,
            "bin": (xbins[:-1] + xbins[1:]) / 2,
            "binlabel": (xbins[:-1] + xbins[1:]) / 2,
        })

    # Convert the list of dictionaries to a DataFrame
    df = pd.DataFrame(histograms)
    
    return df
```

```python
from ma5_expert.histogram.reader import Collection
import matplotlib.pyplot as plt
import numpy as np

# Define the path to the new SAF file
saf_file = '/home/yoxara/Insync/Eff_MA5_Tab8_Cutflows/_run_01_mzp_20_mx_100/Output/SAF/spin1_run_01_mzp_20_mx_100/atlas_exot_2018_06/Histograms/histos.saf'

# Load the SAF file into a Collection object
collection = Collection(original_file=saf_file)

# Get the available histogram names
histo_names = collection.histo_names

# Check if any histograms are available
if histo_names:
    # Select the first histogram to plot
    histogram_name = histo_names[0]
    
    # Get histogram data
    xbins, bins, weights = collection.get_histogram(histogram_name)
    
    # Create the histogram plot
    plt.figure(figsize=(10, 6))
    plt.hist(xbins[:-1], bins=xbins, weights=weights, histtype='step', label=histogram_name)
    plt.xlabel('Bin')
    plt.ylabel('Value')
    plt.title(f'Histogram: {histogram_name} from histos.saf')
    plt.legend()
    plt.show()
else:
    print("No histograms found in the file.")
```
