```python
import ma5_expert as ma5
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
sns.set(style="white", context="paper", font_scale=1.8, 
        rc={"text.usetex": True, 
            'font.family':'Times New Roman', 
            'font.serif':'Times New Roman', 
            "savefig.dpi": 300})
plt.rcParams['text.usetex'] = True

# Configure the plot aesthetics and labels
def configure_axes(ax, info=None):   
    if info is not None:
        multiline_info = "\n".join(info)
        ax.text(0.2, 0.5, multiline_info, transform=ax.transAxes, fontsize=16,
                color='black', bbox=dict(facecolor='silver', alpha=0.5,
                                         edgecolor='silver', boxstyle='round,pad=0.8',
                                         linewidth=0.2))
    def set_x_label(label, size=18, labelpad=5):
        ax.set_xlabel(label, size=size, labelpad=labelpad, loc="right", fontname='serif')
    def set_y_label(label, size=18, labelpad=5):
        ax.set_ylabel(label, size=size, labelpad=labelpad, loc="top", fontname='serif')
    def set_legend(framealpha=0.5, fontsize='16'):
        ax.legend(framealpha=framealpha, loc='best', fontsize=fontsize)
    ax.set_x_label = set_x_label
    ax.set_y_label = set_y_label
    ax.minorticks_on() 
    ax.tick_params(direction='in', length=10, width=1.1, colors='black', which='major', top=True, right=True)
    ax.tick_params(direction='in', length=5, width=0.6, colors='black', which='minor', top=True, right=True)
    ax.xaxis.set_ticks_position('both')
    ax.yaxis.set_ticks_position('both')
    for spine in ['left', 'right', 'top', 'bottom']:
        ax.spines[spine].set_linewidth(1.2)
    sns.despine(left=False, bottom=False, right=False, top=False, offset=None, trim=False)
    ax.tick_params(axis='x', labelcolor='black', labelsize=16)
    ax.tick_params(axis='y', labelcolor='black', labelsize=16)

# Define the path for MadAnalysis and the analysis
ma5_path = "../../MA5"
ma5.BackendManager.set_madanalysis_backend(ma5_path)
```

# Histogram Collection


```python
# Create the histogram collection
collection = ma5.histogram.Collection(
    "/home/yoxara/Insync/SUSY_MA5/_run_01_mt1_450_mneu1_443/Output/SAF/susy_run_01_mt1_450_mneu1_443/atlas_exot_2018_06/Histograms/histos.saf",
    xsection=2.634011e-28, lumi=139.
)
# Print the histogram collection
print(collection)
```

```python
# Get the string representation of the collection
collection_str = str(collection)

# Extract histogram names by parsing the string
histogram_names = []
for line in collection_str.splitlines():
    if "*" in line:
        # Extract the histogram name, which is after the "* " and before the ":"
        histogram_name = line.split(":")[0].split("*")[1].strip()
        histogram_names.append(histogram_name)

# Print the extracted histogram names
print("Extracted histogram names:", histogram_names)
```

```python
#To plot each histogram, we'll access them by known names manually
histogram_list = ["MET_histo"]#histogram_names
#[
#    "MET_histo", "MET_EM0", "MET_EM1", "MET_EM2", "MET_EM3", "MET_EM4", 
#    "MET_EM5", "MET_EM6", "MET_EM7", "MET_EM8", "MET_EM9", "MET_EM10", 
#    "MET_EM11", "MET_EM12"
#]
```

```python
# Define a dictionary for custom names
custom_names = {
    "MET_histo": "Total MET",
  #  "MET_EM0": "MET bin 200-250 GeV",
   # "MET_EM1": "MET bin 250-300 GeV",
 #   "MET_EM2": "MET bin 300-350 GeV",
 #   "MET_EM3": "MET bin 350-400 GeV",
 #   "MET_EM4": "MET bin 400-500 GeV",
 #   "MET_EM5": "MET bin 500-600 GeV",
 #   "MET_EM6": "MET bin 600-700 GeV",
 #   "MET_EM7": "MET bin 700-800 GeV",
 #   "MET_EM8": "MET bin 800-900 GeV",
 #   "MET_EM9": "MET bin 900-1000 GeV",
 #   "MET_EM10": "MET bin 1000-1100 GeV",
 #   "MET_EM11": "MET bin 1100-1200 GeV",
 #   "MET_EM12": "MET bin 1200-1300 GeV"
}
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
# Iterate through each known histogram name and plot
for hist_name in histogram_list:
    xbins, bins, weights = collection.lumi_histogram(hist_name)
    #print(f"* {hist_name}: [nbin: {len(bins)-1}, min: {bins[0]}, max: {bins[-1]}]")
    # Normalize the weights so the area under the histogram sums to 1
    total_weights = sum(weights) if sum(weights) != 0 else 1  # Avoid division by zero
    normalized_weights = [w / total_weights for w in weights]
    # Plot the normalized histograms
    plt.hist(xbins, bins=bins, weights=weights, alpha=0.5, label=custom_names.get(hist_name, hist_name))
configure_axes(ax)
ax.set_x_label("$P_{T}\ {\\rm [GeV]}$")
ax.set_y_label("${\\rm Number\ of\ events}$")
ax.set_xlim([200, 1400])
ax.set_yscale('log')
ax.legend(loc='best', fontsize=10)
plt.show()
```
